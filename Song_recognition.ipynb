# 3 Methodology

In this mini-project, the question under investigation is whether a machine learning model is capable of identifying the proper song title in a 10-second audio recording of a human hum or whistle. The issue is difficult as the humming is highly heterogeneous among the users and generally unreliable in both pitch, articulations, tempo, and timbre (Siahaan and Sianipar, 2022). Hummed audio, contrary to full studio recording, does not have organised harmonic content, is noisy, and has poor rhythmic support, which complicates musical feature extraction and classification. The task is similar to the query-by-humming systems of the real world since the idea is to find a song with the help of limited, distorted, user-created sounds.

# 4 Implemented ML prediction pipelines

The pipeline of machine learning that is implemented will convert raw hummed audio to an appropriate format that can be used to classify it and make an intelligent guess about the label of the most probable song. The data is processed by the pipeline in four phases, which include loading raw audio, converting it into structured features, normalisation and dimensionality reduction as well as classification. The data being fed into the system is a 10 seconds .waw file and the output is one of the eight predetermined labels of the songs. The audio waveform (which is originally an unstructured signal in the time domain) is gradually transformed into a compressed numerical representation by stages that use manually devised signal descriptors and eventually to lower-dimensional vectors, through scaling and the PCA. The classification models take these vectors as the input.

## 4.1 Transformation stage

This is because during this stage, the raw waveform is transformed into useful numerical environmental descriptors. The model removes MFCCs with their deltas to capture spectral envelope variation, chroma variation to capture the harmonic variation, spectral centroid, bandwidth, rolloff, contrast, zero-crossing rate, and tempo. This gives a 87 dimensional feature vector. As the classical machine learning models have fixed-length structured inputs, handcrafted features were selected, as these are lightweight, computationally efficient, interpretable, and are frequently utilized in the audio classification problem. Then Principal Component Analysis (PCA) is used to eliminate redundancy and maintain 85 percent of the Schwartz, which creates a 29-dimensional representation to minimize overfitting.

## 4.2 Model stage

Several models were tested, and these were random Forest, grad Boosting, linear SVM and RBF-SVM. The last model that was chosen was the linear Support Vector Maker owing to the fact that it presented the best generalisation, least train-test gap as well as reliable cross-validated performance. It has regularised linear direction that is effective when dealing with data that is small and countries that vary widely across classes.

## 4.3 Ensemble stage

There was no other ensemble other than tree-based models that was implemented. Initial evaluation of Random Forest and Gradient Boosting was done due to their inherent ability to behave as ensembles and the possibility to model non-linear interaction of features but high levels of overfitting and not being chosen as the final technique.

```python
def extract_features_improved(audio_path, duration=10):
    try:
        y, sr = librosa.load(audio_path, duration=duration)

        # MFCCs with deltas
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
        mfccs_delta = librosa.feature.delta(mfccs)

        mfccs_mean = np.mean(mfccs, axis=1)
        mfccs_std = np.std(mfccs, axis=1)
        mfccs_delta_mean = np.mean(mfccs_delta, axis=1)

        # Chroma features with deltas
        chroma = librosa.feature.chroma_stft(y=y, sr=sr)
        chroma_delta = librosa.feature.delta(chroma)

        chroma_mean = np.mean(chroma, axis=1)
        chroma_std = np.std(chroma, axis=1)
        chroma_delta_mean = np.mean(chroma_delta, axis=1)

        # Spectral features
        spectral_centroid = np.mean(librosa.feature.spectral_centroid(y=y, sr=sr))
        spectral_rolloff = np.mean(librosa.feature.spectral_rolloff(y=y, sr=sr))
        spectral_bandwidth = np.mean(librosa.feature.spectral_bandwidth(y=y, sr=sr))
        spectral_contrast = np.mean(librosa.feature.spectral_contrast(y=y, sr=sr), axis=1)

        # Temporal features
        zcr = np.mean(librosa.feature.zero_crossing_rate(y))
        tempo = librosa.beat.tempo(y=y, sr=sr)[0]

        # Combine all features
        features = np.concatenate([
            mfccs_mean, mfccs_std, mfccs_delta_mean,  # 39 features
            chroma_mean, chroma_std, chroma_delta_mean,  # 36 features
            [spectral_centroid, spectral_rolloff, spectral_bandwidth],  # 3 features
            spectral_contrast,  # 7 features
            [zcr, tempo]  # 2 features
        ])

        return features

    except Exception as e:
        print(f"Error processing {audio_path}: {e}")
        return None
```

```python
print("\nExtracting improved features from all audio files...")
features_list = []
labels_list = []
failed_files = []

for idx, row in df.iterrows():
    if (idx + 1) % 50 == 0:
        print(f"Processed {idx + 1}/{len(df)} files...")

    features = extract_features_improved(row['filepath'])
    if features is not None:
        features_list.append(features)
        labels_list.append(row['label'])
    else:
        failed_files.append(row['filepath'])

print(f"\nFeature extraction complete!")
print(f"Successfully processed: {len(features_list)} files")
print(f"Failed: {len(failed_files)} files")
```

2 Problem Formulation  
This mini-project is aimed to create a machine learning system that can recognise the proper song title based on the 10-second audio recording of a human hum or whistle. It is not an easy task to classify humming as it varies a lot in different people, and it is not stable in harmonics and usually is dissimilar to the original pitch or tempo (Siahaan and Sianipar, 2022). In comparison to full musical recordings, hummed audio has less spectral richness and feature extraction and song discrimination are much more difficult. This is an interesting task as it is similar to real-world systems of retrieving music such as query by humming but with constrained, noisy, and user generated audio settings.  

3 Methodology  
The methodology is divided into four major parts, namely dataset preparation, feature extraction, model training, and evaluation. Following the derivation of the dataset in the MLEnd Hums and Whistles II subset, the audio file names were processed to retrieve the actual label of the song. Eight classes of songs were found, and each of the songs was represented by 50 samples (balanced dataset).  
The audio clips were manipulated by using manually designed feature extraction pipeline consisting of MFCCs, chroma vectors, spectral statistics, zero-crossing rate, and tempo (Milojković, 2025). The feature space reduced to 55 dimensions (the result of the reduction) served as the input of a Random Forest classifier.  
To ensure equal proportions of classes, stratified train-validation splitting (80/20) method was used to divide the dataset. The factors used to assess model performance were accuracy, confusion matrix, per-class accuracy, classification report (precision, recall, F1-score), ranking feature importance, and error analysis (Olabanji, 2023). This multi-metric analysis enabled the discovery of the global performance as well as weaknesses that are class-specific.  

4 Implemented ML Prediction Pipelines  
The pipeline of predictions will include the following features: loading audio into it, extracting features, standardisation of features, the classifier, the labelling of the predicted song. The raw waveform audio (.wav) is the input data and one of the eight song categories is the output.  

4.1 Transformation Stage  
Raw audio signals were transformed into ordered numeric signal. MFCCs represent the spectral envelope, chroma characteristics are used to describe the harmonic content and spectral centroid, rolloff, bandwidth, zero-crossing rate, and tempo are other rhythmic and timbral indicators. The results are numerical vectors of 55 dimensions (Patkar et al., 2022). The reason as to why this transformation was selected was that the classical ML models need fixed-length feature vectors, and the descriptors are common in audio classification.  

4.2 Model Stage  
The baseline model should be characterised by a rather robust model, non-linear decision boundaries, and resistance to overfitting on small datasets: such features made a Random Forest classifier the trade-off decision. It also takes advantage of heterogeneous features, and does not need a lot of hyperparameter optimization. The prediction is the resultant actual label.  

4.3 Ensemble Stage  
There was no implementation of any ensemble of independent models, but the owner Random Forest is an ensemble of decision trees. It was selected because it minimised variance when compared to single-tree models.  

5 Dataset  
The data set comprises of 400 hummed and whistled music files in eight songs. The final token was used to derive labels after the extraction of filenames (ex.: TryEverything). The stratified sampling was used to create train and validation sets to provide IID and class-balanced splits. One disadvantage is that the quality of humming is highly diverse, so it brings about a considerable amount of intra-class variance to a basic feature.  

6 Experiments and Results  
The Random Forest had a validation accuracy of 23.75, and great overfitting (100% training accuracy). Confusion matrices demonstrated there were much cross-song confusion especially of the word Married, RememberMe and NewYork. Analysis of feature importance revealed that tempo and chroma statistics dominated, and this implies the model is based on global spectral patterns, and not on melodic shape.  

7 Conclusions  
The +Random Forest approach with handcrafted-feature offers limited predictive capability as a result of the complexity of humming. It can be improved in the future with mel-spectrogram CNNs, pre-trained embeddings (YAMNet/OpenL3) and data augmentation.
